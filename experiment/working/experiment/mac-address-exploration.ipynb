````python
import os 
import sys
import pandas as pd
from typing import Dict, List
````

````python
DATASET_FOLDER = "../../../dataset"
DEVICE_LIST_1 = DATASET_FOLDER +"/identify-iot-devices-and-events-based-on-packet-length-from-encrypted-traffic/device-mapping/List_Of_Devices.txt"
DEVICE_LIST_2_FOLDER = DATASET_FOLDER + "/IoT-Dataset/devices"

sys.path.append(os.path.abspath(DEVICE_LIST_2_FOLDER))

from devices import devices
device_mapping_2 = devices
````

````python
pd.set_option("display.max_rows", None)
pd.set_option("display.max_columns", None)
pd.set_option("display.max_colwidth", None)
pd.set_option("display.width", None)
````

````python
device_mapping1 = []
with open(DEVICE_LIST_1, 'r') as f:
    for line in f:
        line_data = line.strip().split('\t')
        mapping = [data.strip() for data in line_data if data.strip()]
        if mapping and mapping[0].lower() != "list of devices":
            device_mapping1.append((mapping[0], mapping[1]))

devices_list_1 = {}
for device, mac in device_mapping1:
    if device in devices_list_1:
        device = device + ".1"
    devices_list_1[device] = mac

for device in devices_list_1:
    print(f"{device}: {devices_list_1[device]}")    
````

````python
devices_list_2 = {}
mac_key = "ether"
for device in device_mapping_2:
    if mac_key in device_mapping_2[device]:
        devices_list_2[device] = device_mapping_2[device][mac_key]


for device in devices_list_2:
    print(f"{device}: {devices_list_2[device]}")
````

````python
final_mapping = []
for device, mac in devices_list_1.items():
    final_mapping.append({"device": device, "source": "list-1", "mac-address": mac})
for device, mac in devices_list_2.items():
    final_mapping.append({"device": device, "source": "list-2", "mac-address": mac})
````

````python
mapping_df = pd.DataFrame(final_mapping)
mapping_df.sort_values(by="mac-address")
````

````python
OUI_MAPPING_FILE = DATASET_FOLDER + "/oui-database/oui-ieee-mapping-list.csv"
````

````python
def load_ieee_mapping(ieee_database_file: str) -> Dict[str,List[Dict[str, str]]]:
    ieee_database_df = pd.read_csv(ieee_database_file)
    display(ieee_database_df.head(10))

    ieee_database = {}
    for _, row in ieee_database_df.iterrows():
        prefix = row["Assignment"]

        if prefix in ieee_database:
            print(f"WARNING: Duplicated mapping for {prefix}:\n{row.to_dict()}\n{ieee_database[prefix]}")
            ieee_database[prefix].append({
                "company": row["Organization Name"],
                "registry": row["Registry"],
                "address": row["Organization Address"],
            })
        else:
            ieee_database[prefix] = [{
                "company": row["Organization Name"],
                "registry": row["Registry"],
                "address": row["Organization Address"],
            }]
    return ieee_database
````

````python
oui_ieee_mapping = load_ieee_mapping(OUI_MAPPING_FILE)
````

````python
mapping_df["mac-prefix"] = (
    mapping_df["mac-address"]
    .str.replace(":", "", regex=False)
    .str.upper()
    .str[:6]
)
mapping_df["vendor"] = (
    mapping_df["mac-prefix"].map(
        lambda prefix: [
            record["company"]
            for record in oui_ieee_mapping.get(prefix, [])
        ]
    )
)
````
